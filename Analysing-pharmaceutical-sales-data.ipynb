```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "Student.csv"
data = pd.read_csv(url)
```

```python
sns.scatterplot(x="study_hours", y="marks", data=data)
study_marks_corr = data["study_hours"].corr(data["marks"])
print(study_marks_corr)
plt.show()
```
