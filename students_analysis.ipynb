```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "Student.csv"
data = pd.read_csv(url)
properties = {"font":"Jetbrains Mono", "size":"11", "weight":"bold"}
```

```python
sns.scatterplot(x="study_hours", y="marks", data=data)
study_marks_corr = data["study_hours"].corr(data["marks"])
print(study_marks_corr)
plt.title("Study Hours vs Marks", **properties)
plt.show()
```

```python
bounds = [2,4, 5, 6, 7, 8, 9, 10]
bound_labels = ["3-4","4-5","5-6","6-7","7-8","8-9","9-10"]
data["sleep_bin"] = pd.cut(data["sleep_hours"], bins=bounds, labels=bound_labels)
sleep_marks = data.groupby("sleep_bin")["marks"].mean()
sleep_marks.plot(kind="line", marker="o")
plt.title("Sleep Hours To Marks Graph", **properties, color="darkgreen")
plt.ylabel("Average Marks", **properties, )
plt.xlabel("Sleep Hours", **properties)
print(sleep_marks)
```

```python
sns.scatterplot(x="sleep_hours",y="marks",data=data)
```

```python
print(data["sleep_hours"].corr(data["marks"]))
```

```python
sns.regplot(x="sleep_hours", y="marks",data=data, color="green")
plt.title("Sleep Hours Vs Marks", **properties, color="blue")
plt.ylabel("Average Marks", **properties, )
plt.xlabel("Sleep Hours", **properties)
```

```python
figure, ax = plt.subplots(1,2, figsize=(11,5), sharex=True, sharey=True)
sns.scatterplot(x="screen_time",y="marks",data=data, ax=ax[0])
sns.regplot(x="screen_time",y="marks",data=data, ax=ax[1])
ax[0].set_title("Screen Time Vs Marks", **properties)
ax[1].set_title("Screen Time Vs Marks", **properties)
print(data["screen_time"].corr(data["marks"]))
```

```python
figure, ax = plt.subplots(1,2, figsize=(11,5), sharex=True, sharey=True)
sns.scatterplot(x="gaming_hours",y="marks",data=data, ax=ax[0])
sns.regplot(x="gaming_hours",y="marks",data=data, ax=ax[1])
ax[0].set_title("Gaming Hours Vs Marks", **properties)
ax[1].set_title("Gaming Hours Vs Marks", **properties)
print(data["gaming_hours"].corr(data["marks"]))
```

```python
figure, ax = plt.subplots(1,2, figsize=(11,5), sharex=True, sharey=True)
sns.scatterplot(x="gaming_hours",y="sleep_hours",data=data, ax=ax[0])
sns.regplot(x="gaming_hours",y="sleep_hours",data=data, ax=ax[1])
ax[0].set_title("Gaming Hours Vs Sleep", **properties)
ax[1].set_title("Gaming Hours Vs Sleep", **properties)
print(data["gaming_hours"].corr(data["sleep_hours"]))
```

```python
figure, ax = plt.subplots(1,2, figsize=(11,5), sharex=True, sharey=True)
sns.scatterplot(x="gaming_hours",y="screen_time",data=data, ax=ax[0])
sns.regplot(x="gaming_hours",y="screen_time",data=data, ax=ax[1])
ax[0].set_title("Gaming Hours Vs Screen Time", **properties)
ax[1].set_title("Gaming Hours Vs Screen Time", **properties)
print(data["gaming_hours"].corr(data["screen_time"]))
```

```python
marks_corr = data.select_dtypes("number").corr()["marks"]
marks_corr = marks_corr.sort_values(ascending=False)
plt.figure(figsize=(11,5))
sns.barplot(marks_corr)
plt.title('Biggest Factor Affecting Marks(Ranked)', **properties)
```

```python
sns.heatmap(data.select_dtypes("number").corr(), annot=True, cmap='coolwarm', xticklabels=['Study Hours', 'Sleep Hours', 'Screen Time', 'Gaming Hours', 'Attendance Percent', 'Stress Level', 'Marks']
,yticklabels=['Study Hours', 'Sleep Hours', 'Screen Time', 'Gaming Hours', 'Attendance Percent', 'Stress Level', 'Marks']
)
plt.show()
```

```python
sns.histplot(data["marks"])
plt.title("Distribution of Marks", **properties)
```

```python

bins = [0,2,5,data["gaming_hours"].max()]
labels=["low","moderate","high"]
data["gaming_category"] = pd.cut(data["gaming_hours"],bins=bins, labels=labels)
sns.boxplot(y="marks",x="gaming_category",data=data)
plt.xlabel("Gaming Category")
plt.ylabel("Marks")
plt.title("Marks Distribution By Gaming Cateogory", **properties)
```
