```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "Student.csv"
data = pd.read_csv(url)
properties = {"font":"Jetbrains Mono", "size":"11", "weight":"bold"}
print(data.columns)
```

```python
bounds = [2,4, 5, 6, 7, 8, 9, 10]
bound_labels = ["3-4","4-5","5-6","6-7","7-8","8-9","9-10"]
data["sleep_bin"] = pd.cut(data["sleep_hours"], bins=bounds, labels=bound_labels)
sleep_marks = data.groupby("sleep_bin")["marks"].mean()
sleep_marks.plot(kind="line", marker="o")
plt.title("Sleep Hours To Marks Graph", **properties, color="darkgreen")
plt.ylabel("Average Marks", **properties, )
plt.xlabel("Sleep Hours", **properties)
print(sleep_marks)
```
